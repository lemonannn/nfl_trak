---------------------------------------
COLLECT VALID IMAGES
---------------------------------------

```python
def collect_images(video_dir):
    images = []

    for ext in IMAGE_EXTS:
        images.extend(glob.glob(os.path.join(video_dir, ext)))
    print(video_dir)

    valid = []

    for img_path in images:
        label_path = os.path.splitext(img_path)[0] + ".txt"

        if os.path.exists(label_path):
            valid.append(img_path)

    return sorted(valid)
```

```python
train_images = []
val_images = []
```

```python
for vid in train_videos:
    train_images.extend(collect_images(vid))
```

```python
for vid in val_videos:
    val_images.extend(collect_images(vid))
```

---------------------------------------
WRITE SPLITS
---------------------------------------

```python
with open("train.txt", "w") as f:
    for img in train_images:
        f.write(img + "\n")
```

```python
with open("val.txt", "w") as f:
    for img in val_images:
        f.write(img + "\n")
```

```python
import cv2
!pip install -q ultralytics
from ultralytics import YOLO

# ----------------------------
# LOAD MODEL
# ----------------------------
model = YOLO("/content/drive/MyDrive/nfl_trak/v4/weights/best.pt")

print("Classes:", model.names)

# ----------------------------
# OPEN VIDEO
# ----------------------------
cap = cv2.VideoCapture("test13.mp4")

if not cap.isOpened():
    raise RuntimeError("Could not open input video")

fps = cap.get(cv2.CAP_PROP_FPS)
w = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
h = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))

if fps == 0:
    fps = 30

# ----------------------------
# OUTPUT VIDEO
# ----------------------------
out = cv2.VideoWriter(
    "/content/drive/MyDrive/nfl_trak/output_annotated.mp4",
    cv2.VideoWriter_fourcc(*"mp4v"),
    fps,
    (w, h)
)

# ----------------------------
# STATS
# ----------------------------
frame_count = 0
player_counts = []

# ----------------------------
# PROCESS VIDEO
# ----------------------------
while True:
    ret, frame = cap.read()

    if not ret:
        break

    frame_count += 1
    player_count = 0

    # Explicit settings
    results = model(
        frame,
        conf=0.50,
        imgsz=960,
        verbose=False
    )[0]

    for box in results.boxes:
        conf = float(box.conf[0])

        x1, y1, x2, y2 = box.xyxy[0].tolist()
        x1, y1, x2, y2 = map(int, (x1, y1, x2, y2))

        player_count += 1

        # Center point
        cx = (x1 + x2) // 2
        cy = (y1 + y2) // 2

        # Bounding box
        cv2.rectangle(
            frame,
            (x1, y1),
            (x2, y2),
            (0, 255, 0),
            2
        )

        # Center marker
        cv2.circle(
            frame,
            (cx, cy),
            3,
            (0, 255, 0),
            -1
        )

        # Confidence
        cv2.putText(
            frame,
            f"{conf:.2f}",
            (x1, y1 - 5),
            cv2.FONT_HERSHEY_SIMPLEX,
            0.5,
            (0, 255, 0),
            1
        )

    player_counts.append(player_count)

    # Detection count overlay
    cv2.putText(
        frame,
        f"Players: {player_count}",
        (20, 40),
        cv2.FONT_HERSHEY_SIMPLEX,
        1,
        (0, 255, 0),
        2
    )

    out.write(frame)

# ----------------------------
# CLEANUP
# ----------------------------
cap.release()
out.release()

# ----------------------------
# SUMMARY STATS
# ----------------------------
if player_counts:
    avg_players = sum(player_counts) / len(player_counts)

    print("\n----- SUMMARY -----")
    print(f"Frames processed: {frame_count}")
    print(f"Average detections/frame: {avg_players:.2f}")
    print(f"Minimum detections/frame: {min(player_counts)}")
    print(f"Maximum detections/frame: {max(player_counts)}")

print("\nSaved: output_annotated.mp4")
```
